## 2️⃣ Looping with Index (range + len)

Use Case: Seating Allocation System

Used when position or index matters.

```python
students = ["Rahim", "Karim", "Sadia", "Bipul", "Hasan"]

# range(start, end, step)

for i in range(len(students)):
    print(f"Seat No: {i+1} allocated to {students[i]}")

print(f"\n{students[-1]}, {students[-2]}, {students[-3]}\n")

for i in range(1, len(students), 2):
    print(f"Seat No: {i+1} allocated to {students[i]}")

# print(len(students)-1 == 3)
```

```python
# Reverse Print

students = ["Rahim", "Karim", "Sadia"]

for i in range(len(students)-1 , 0, -1): # Always stops before the ending point. In this case, the ending point is: 0, and the starting point is 2, and it going backward through -1 step. So it will start just before 0, which is 1.
   print(students[i])

print("\n")

for i in range(len(students)-1 , -1, -1):
   print(students[i])
```

## 3️⃣ Looping with enumerate() (Best Practice)

```python
tasks = ["Design UI", "Build API", "Write Tests","Model Training","Deployment"]


# enumerate(tasks) or enumerate(tasks, sgtart = 0)
for index, task in enumerate(tasks):
    print(f"Task {index}: {task}")

print("\n")


# enumerate(tasks, start = 1):
for index, task in enumerate(tasks, start = 1):
    print(f"Task {index}: {task}")
```

## 2. Nested Loops – E-commerce Outfit Generator

```python
shirts = ["Blue Denim", "White Linen", "Black Cotton"]
pants = ["Khaki Chinos", "Black Jeans"]

outfits = []

for shirt in shirts:
    for pant in pants:
        outfits.append(f"{shirt} + {pant}")
        # print(f"{shirt} + {pant}\n")

print(f"Printing as lists: {outfits}\n")
print(f"Printing as individual outfits:\n")
for combo in outfits:
    print(combo)
```

## 3. Break & Continue – Payment Processing Pipeline

These statements give you finer control over your loops.

    break: Stops the loop entirely, even if the loop condition is still true.

    continue: Skips the current iteration and moves to the next item in the sequence. Starts from beginning of loop.

## Problem Statement

You are designing a basic file scanning system for a computer.
The system scans files one by one from a given list and follows these rules:

1. System files with the extension `.sys` must be skipped and not scanned.

2. If a critical threat file named `"virus.exe"` is detected, the scan must stop immediately.

3. All other files should be scanned and marked as Clean.

The list of files to be scanned is:

`files = ["photo.jpg", "system.sys", "virus.exe", "document.pdf", "malware.exe"]`

##Tasks:

a) Write a Python program to implement the above scanning rules using a for loop.

b) Use the `continue` statement to skip system files.

c) Use the `break` statement to immediately stop the scan when a critical threat is detected.

d) Display appropriate messages for:

1. Skipped system files

2. Critical threat detection

3. Successfully scanned clean files



```python
files = ["photo.jpg", "system.sys", "virus.exe", "document.pdf", "malware.exe"]

for file in files:
    if file.endswith(".sys"):
        print(f"Skipping {file} (system file)")
        continue

    if file.endswith(".exe"):
        print(f"Critical threat detected: {file}. Shutting down scan!")
        break

    print(f"Scanning {file}....Clean")

```

## Problem Statement

You are developing a transaction processing system for an online platform.
Each transaction contains:

1. a unique transaction ID

2. a processing status

The system must process transactions based on the following rules:

1. Pending transactions must be skipped without any processing.

2. If a critical database error `("Error_DB_Down")` is detected, the system must stop immediately.

3. All other transactions should be processed normally.

4. The system must keep track of the total number of successfully processed transactions.

The list of transactions is given below:

`transactions = [
    {"id": 101, "status": "Success"},
    {"id": 102, "status": "Pending"},
    {"id": 103, "status": "Success"},
    {"id": 104, "status": "Error_DB_Down"},
    {"id": 105, "status": "Success"}
]`

##Tasks

a) Write a Python program to process the transactions following the rules above.

b) Use the `continue` statement to skip pending transactions.

c) Use the `break` statement to halt the system when a critical error occurs.

d) Print a message when each transaction is successfully processed.

e) Display the total number of processed transactions at the end.

## Expected Output

`Transaction 101 processed`

`Transaction 103 processed`

`🚨 Critical error detected. Halting system.`

`Total processed transactions: 2`



```python
transactions = [
    {"id": 101, "status": "Success"},
    {"id": 102, "status": "Pending"},
    {"id": 103, "status": "Success"},
    {"id": 104, "status": "Error_DB_Down"},
    {"id": 105, "status": "Success"}
]

counter = 0

for transaction in transactions:
    if transaction["status"] == "Pending":
        continue

    if transaction["status"] == "Error_DB_Down":
        print(f"🚨 Critical error detected. Halting system.")
        break
    print(f"tansaction {transaction["id"]} processed")
    counter += 1

print(f"Total processed transactions: {counter}")
```

## 5. List Comprehension – Data Transformation




### When, Where, and Why to Use It

List comprehension is a **clean, expressive, and Pythonic** way to create new lists from existing data. It is widely used in **professional Python codebases** for data processing, filtering, and transformation.



### ✅ WHEN to Use List Comprehension

Use list comprehension when you want to:

- Create a **new list** from another list or iterable
- **Transform** each element (e.g., formatting, calculations)
- **Filter** data based on a condition
- Write **short, readable, single-purpose logic**
- Replace a simple `for` loop that only appends values


```python
# Using Traditional Loop
raw_names = ["  alice", "BOB ", " chArlie ", "David"]
clean_name=[]
for name in raw_names:
    clean_name.append(name.strip().capitalize()) # append
print(clean_name)
```

```python
# Using List Comprehension
raw_names = ["  alice", "BOB ", " chArlie ", "David"]

clean_names= [name.strip().capitalize() for name in raw_names]
print(clean_names)
```

```python
# filtering Using Traditional Loop
order_amounts = [120, 550, 45, 1200, 300, 750]
vip_orders = []
for i in order_amounts:
   if i > 500:
      vip_orders.append(i)
      print(i)

print(vip_orders)
```

```python
# filtering Using List Comprehension
order_amounts = [120, 550, 45, 1200, 300, 750]
vip_orders1 = [i for i in order_amounts if i > 500]
print(vip_orders1)

vip_orders2 = [i  if i > 500 else 0 for i in order_amounts]
print(vip_orders2)
```

```python
raw_names = ["  alice", "BOB ", " chArlie ", "David"]
clean_names = [name.strip().capitalize() for name in raw_names]

order_amounts = [120, 550, 45, 1200, 300, 750]
vip_orders = [amt for amt in order_amounts if amt > 500]

print("Clean Names:", clean_names)
print("VIP Orders:", vip_orders)
```

### Benefits of List Comprehensions

List comprehensions offer several advantages over traditional `for` loops for creating lists:

1.  **Conciseness**: They allow you to create new lists in a single, compact line of code, reducing the overall lines of code and making it easier to read and understand the intent.

2.  **Readability**: Once familiar with the syntax, list comprehensions are often more readable than `for` loops for simple list construction and transformation tasks, as they express the entire operation in a clear, declarative way.

3.  **Performance**: In many cases, list comprehensions can be slightly faster than equivalent `for` loops, especially for larger datasets. This is because they are optimized at the C level in Python's interpreter.

4.  **Reduced Boilerplate**: They eliminate the need for `append()` calls and explicit loop initialization, leading to less boilerplate code.

### A. Single Item Transformation

**Use Case: Cleaning User Input** When users sign up for your website, they often type their names with messy capitalization. You need to standardize them for your database.

```python
raw_usernames = [
    "  john_doe ",
    "SARA99",
    " miKe_tySon ",
    "admin"
]

# --- Using List Comprehension (Recommended) ---
normalized_lc = [
    username.strip().lower()
    for username in raw_usernames
]

# --- Using a Traditional For Loop ---
normalized_loop = []

for username in raw_usernames:
    cleaned = username.strip().lower()
    normalized_loop.append(cleaned)

print("List Comprehension Output:", normalized_lc)
print("For Loop Output:", normalized_loop)
```

### B. Filtering Data

**Use Case: Finding Premium Customers** From a list of order totals, extract only the "High Value" orders (over $500) to send them a VIP discount code.

```python
daily_orders = [120, 550, 45, 1200, 300, 750]
VIP_THRESHOLD = 500

# --- Using List Comprehension (Recommended) ---
vip_orders_lc = [
    order for order in daily_orders
    if order >= VIP_THRESHOLD
]

# --- Using a Traditional For Loop ---
vip_orders_loop = []

for order in daily_orders:
    if order >= VIP_THRESHOLD:
        vip_orders_loop.append(order)

print("VIP Orders (List Comprehension):", vip_orders_lc)
print("VIP Orders (For Loop):", vip_orders_loop)
```

### C. Pair/Tuple Generation (Advanced)

**Use Case: Inventory Value Calculator** You have a list of stock counts and you want to create a list of pairs (tuples) containing the item name and its total value (Quantity × Price).

```python
# Using Traditional Loop

item_names = ["Laptop", "Mouse", "Monitor"]
stock_counts = [10, 50, 20]
price_per_unit = 1000

inventory_list = []

for i in range(len(item_names)):
    inventory_list.append((item_names[i], stock_counts[i] * price_per_unit))
    # inventory_list.append(f"{item_names[i]}, {stock_counts[i] * price_per_unit}")

print(inventory_list)
# print(type(inventory_list))
```

```python
# Using List Comprehensions

item_names = ["Laptop", "Mouse", "Monitor"]
stock_counts = [10, 50, 20]
price_per_unit = 1000

inventory_list = [(item_names[i], stock_counts[i] * price_per_unit) for i in range(len(item_names))]
print(inventory_list)
```


## 2️⃣ Type Conversion (Type Casting)

Python cannot calculate using text.
We convert text into numbers using:
- int()
- float()


```python
# ✅ Easy Example
num1 = input("Enter a number: ")
num2 = input("Enter another number: ")

sum_val = int(num1) + int(num2) # Renamed 'sum' to 'sum_val' to avoid shadowing built-in sum()
print(f"The sum is {sum_val}")
```

```python
num = int(input("Enter a number: ")) #12%2 = 0 ,13%2=  1
rem = num % 2
if rem == 0:
  print("Even")
else:
  print("Odd")
```

```python
# 🧪 Real-Life Example: Pocket Money
money = float(input("How much pocket money do you get? "))
days = int(input("For how many days? "))

print(f"Total money: {money * days}")
```

```python
# ✅ Example: Age Calculator
birth_year = input("Enter your birth year: ")
current_year = 2026
age = current_year - int(birth_year)

print("Your age is", age)
```

```python
# ✅ Example: BMI Calculator

height = 1.78 #(in Inches)
weight = 65.00 #(in KGs)

bmi = weight / (height ** 2)
print(f"Your BMI is {bmi:.3f}")
```

Errors are normal.
Debugging means finding and fixing mistakes.


```python
# 🚨 Common Errors You Will See

# # ❌ TypeError Example 1: Adding a string and an integer
# age = input("Enter age: ")
# print(age + 1) # This line would cause a TypeError if uncommented

# ✅ Fix for TypeError Example 1:
age_type_error_fix = int(input("Enter age (for TypeError fix): "))
print(age_type_error_fix + 1)
```


## 🧮 Sample Project: Simple Calculator


```python
# ✅ Calculator Code (Beginner Friendly)

print("Simple Calculator")
print("Operations: +  -  *  /")

# Take inputs
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Choose an operator (+, -, *, /): ") ## *

# Logic
if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    if num2 == 0:
        result = "Error! Division by zero."
    else:
        result = num1 / num2
else:
    result = "Invalid operator"

# Output
print(f"Result: {result}")

# 🧪 Example Run
# Enter first number: 10
# Enter second number: 5
# Choose an operator (+, -, *, /): *
# Result: 50.0
```

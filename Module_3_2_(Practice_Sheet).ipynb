# Advanced Python Looping Techniques

This notebook covers advanced looping techniques in Python, moving beyond simple `for i in range()` to real-world data manipulation.

These techniques are essential for **Data Science**, **Automation**, and **Backend Development**.

## 1. Iterating Over Basic Data Structures

Lists and Tuples are ordered, Sets are unordered, and Dictionaries require special methods.

```python
# --- Lists & Tuples (Ordered) ---
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"Processing fruit: {fruit}")

# --- Sets (Unordered & Unique) ---
user_ids = {101, 102, 105, 101, 103}
for uid in user_ids:
    print(f"Sending notification to User ID: {uid}")

# --- Dictionaries ---
api_config = {
    "timeout": 30,
    "retries": 3,
    "version": "v2.1"
}

for key, value in api_config.items():
    print(f"Config Setting -> {key}: {value}")
```

## 2. Power Tools: enumerate(), zip() and range()

`zip()` function is used to iterate over multiple lists simultaneously by combining their elements position-by-position into pairs (or tuples) without using indexing.

`range()`is used to generate a sequence of numbers (usually indices) so you can iterate a specific number of times or access list elements using their index.

`enumerate()`is used to iterate over a sequence while simultaneously accessing both the index and the element in each iteration.


```python
# range example 1

userid = [1, 2]
user_name = ["Tahmid", "Tawhid"]
user_email = ["user1@gmail.com", "user2@gmail.com"]
user_active_status = ["True", "False"]

# i=0 -> userid = [1, 2], user_name = ["Tahmid", "Tawhid"], user_email = ["user1@gmail.com", "user2@gmail.com"], user_active_status = ["True", "False"]
# i=1 -> userid = [1, 2], user_name = ["Tahmid", "Tawhid"], user_email = ["user1@gmail.com", "user2@gmail.com"], user_active_status = ["True", "False"]

for i in range(len(userid)):
  print(f"{userid}, {user_name}, {user_email}, {user_active_status}")
```

```python
# range example 2

userid = [1, 2]
user_name = ["Tahmid", "Tawhid"]
user_email = ["user1@gmail.com", "user2@gmail.com"]
user_active_status = ["True", "False"]

for i in range(len(userid)):
  print(f"{userid[i]}: {user_name[i]}, {user_email[i]}, {user_active_status[i]}")
```

```python
# Recap of List Comprehension (without loops)

animals = ['lion', 'tiger', 'monkey', 'elephant', 'frog']
filtered_animals = [animal.title() for animal in animals]

print(f"Without loops: {filtered_animals}")

# with loops

animals = ['lion', 'tiger', 'monkey', 'elephant', 'frog']

filtered_animals_1 = []

for animal in animals:
  filtered_animals_1.append(animal.title())

print(f"With loops:    {filtered_animals_1}")

```

```python
# Enumerate Example
top_sales_reps = ["Alice", "Bob", "Charlie", "David"]

print("Sales Leaderboard:")
for rank, name in enumerate(top_sales_reps, start=1):
    print(f"{rank}. {name}")
```

```python
stock = [15, 0, 42]
## "out of stock" / "in stock"
status = [  "in stock" if i > 0 else "out of stock" for i in stock]
print(status)
```

```python
# zip example 1

userid = [1, 2]
user_name = ["Tahmid", "Tawhid"]
user_email = ["user1@gmail.com", "user2@gmail.com"]
user_active_status = ["True", "False"]

# print(list(zip(userid, user_name, user_email, user_active_status)))

zipped = zip(userid, user_name, user_email, user_active_status)

for id, name, email, active_status in zipped:
  print(id, name, email, active_status)
```

```python
# Zip Example 2
products = ["Laptop", "Mouse", "Keyboard"]
stock = [15, 0, 42]
prices = [1200, 25, 75]

for item, qty, price in zip(products, stock, prices):
    status = "In Stock" if qty > 0 else "Out of Stock" # one line
    print(f"Product: {item} | Price: ${price} | Status: {status} ({qty} units)")
```

```python
# zip Example 3

# Real-world Backend Use Case for zip()
# Imagine you're processing user data from a database or API

user_ids = ["user_001", "user_002", "user_003"]
user_names = ["Alice", "Bob", "Charlie"]
user_emails = ["alice@example.com", "bob@example.com", "charlie@example.com"]
user_active_status = [True, False, True]

print("\nUser Account Summary:")
for uid, name, email, active in zip(user_ids, user_names, user_emails, user_active_status):
    status_text = "Active" if active else "Inactive"
    print(f"ID: {uid}, Name: {name}, Email: {email}, Status: {status_text}")
```

```python
# --- Length of zip() with different list items ---
# zip() stops when the shortest iterable is exhausted.
list1 = [1, 2, 3, 4]
list2 = ['a', 'b', 'c']
list3 = [True, False]

print("\nZipping lists of different lengths:")
for item1, item2, item3 in zip(list1, list2, list3):
    print(f"({item1}, {item2}, {item3})")

print(f"The length of zipped result is always determined by the shortest iterable, in this case, {len(list3)}.")
```

```python
# zip() with Sets (Order is not guaranteed due to set's nature)
set_colors = {"red", "blue", "green"}
set_hex_codes = {"#FF0000", "#0000FF", "#008000"}
print("\nZipping sets (order is arbitrary):")
for color, hex_code in zip(set_colors, set_hex_codes):
    print(f"Color: {color}, Hex: {hex_code}")
```

```python
# zip() with Dictionary keys (iterates over keys by default)
dict_students = {"Math": 90, "Science": 85}
dict_teachers = {"Math": "Mr. Smith", "Science": "Ms. Jones"}
print("\nZipping dictionary keys:")
for subject, teacher in zip(dict_students.keys(), dict_teachers.values()):
    print(f"Subject: {subject}, Teacher: {teacher}")
```

```python
# zip() with Tuples
tuple_names = ("David", "Eve")
tuple_ages = [28]
print("\nZipping tuples:")
for name, age in zip(tuple_names, tuple_ages):
    print(f"Name: {name}, Age: {age}")
```

```python
# If you want to zip dictionary items (key-value pairs):
dict_cities = {"London": "UK", "Paris": "France"}
dict_populations = {"London": 9_000_000, "Paris": 2_000_000}
print("\nZipping dictionary items (key-value pairs):")


for (city1, country), (city2, population) in zip(dict_cities.items(), dict_populations.items()):
    # Note: If keys are the same, you can just use one key
    print(f"City: {city1}, Country: {country}, Population: {population}")
```

## Task 1: Premium Product Classification

Create a new list called `premium_products` using list comprehension:

If a product's price is greater than 50 → keep the product name.

Otherwise → replace it with `"Standard Product"`.

Expected Output:

`Premium items: ['Laptop', 'Standard Product', 'Keyboard']`

##Task 2: Inventory Summary

Create a dictionary called stock_map using dictionary comprehension:

If stock quantity is greater than 0 → value should be "Available".

Otherwise → value should be "Empty".

Expected Output:

`Inventory Summary: {'Laptop': 'Available', 'Mouse': 'Empty', 'Keyboard': 'Available'}`



```python
products = ["Laptop", "Mouse", "Keyboard"]
stock = [15, 0, 42]
prices = [1200, 25, 75]

premium_products = [product_name if product_price > 50 else "Standard Product" for (product_name, stock_quantity, product_price) in zip(products, stock, prices)]
print(f"Premium Items: {premium_products}")

stock_map = {product_name: ("Available" if stock_quantity > 0 else "Empty") for (product_name, stock_quantity, product_price) in zip(products, stock, prices)}
print(f"Inventory Summary: {stock_map}")
```

```python
# Problem 1: Order Total Calculator

quantities = [2, 5, 1]
unit_prices = [10.5, 2.0, 100.0]

total = 0
for q, p in zip(quantities, unit_prices):
    total += q * p
    print(total)

print(f"Total Order Cost: ${total}")
```

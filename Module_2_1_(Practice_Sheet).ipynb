## 4. ATM PIN Verification System Using `while` loop


---
Question

A bank ATM system allows a user to enter a **4-digit PIN** to access their account.  
The system follows the rules below:

- The correct PIN is **"1234"**.
- The user is allowed a **maximum of 3 attempts** to enter the correct PIN.
- If the user enters the correct PIN within 3 attempts, the system displays  
  **"Access Granted"** and stops asking for the PIN.
- If the user fails to enter the correct PIN after 3 attempts, the system displays  
  **"Card Blocked"**.
- After the PIN verification process ends, the program prints  
  **"Contact with customer care"**.


---

## Task

Write a **Python program** to implement it. You can customize the answer to make it more user friendly. But the core answer must be present.


```python
correct_pin = "1234"

attempt = 0;

while attempt < 3:
    pin_number = input("Enter your 4 digit pin number: ")

    if pin_number == correct_pin:
        print("\nAccess Granted")
        break
    else:
        if attempt == 2:
          print("Your Card is blocked.\nContact with customer care!")
        else:
          print(f"Wrong input....! Try again within {(3 - 1) - attempt} attempts\n")
        attempt += 1
```

# 5. Mini Project: Smart User Validator

## Question

You are given a raw user data string coming from a system log.  
The data may contain **extra spaces**, **mixed letter cases**, and **multiple fields** separated by special characters.

### Raw Data Format

"   USER:2025-Ostad-77 | STATUS:active "

Each log entry contains:
- A **USER ID**, prefixed by `USER:`
- A **STATUS**, prefixed by `STATUS:`
- The fields are separated by the `|` symbol

---

## Validation Rules

A user is considered **authorized** if **both** of the following conditions are satisfied:

1. The user status is **ACTIVE** (case-insensitive).
2. The user ID starts with the year **2025**.

---

## Task

Write a Python program named **Smart User Validator** that performs the following steps:

1. Removes extra spaces from the raw data.
2. Converts all characters to **uppercase**.
3. Splits the data using the `|` character.
4. Extracts:
   - the user ID  
   - the user status  
   - the year from the user ID
5. Validates the user based on the given rules.
6. Displays:
   - `User <user_id> authorized` if the user is valid  
   - `Access denied` otherwise

---

## Expected Output (for the given input)

User 2025-OSTAD-77 is authorized

---

### Constraints

- Use built-in string functions only.
- Do not hard-code extracted values.
- The program should work for similar input formats.

```python
raw_data = "   USER:2025-Ostad-77 | STATUS:active   "
clean_data = raw_data.strip().upper()
# print(clean_data)

sections = clean_data.split("|")
# print(sections)

user_id = sections[0].split(":")[1]
# print(user_id)
# print(type(user_id))

status = sections[1].strip().split(":")[1]
# print(status)

year = user_id[0:4]
# print(year)

if status == "ACTIVE" and year == "2025":
  print(f"User {user_id}is authorized")
else:
  print("Access denied")
```
